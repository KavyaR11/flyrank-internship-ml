## 2. The question: decision, action, cost of a wrong call


Unit of analysis: one page

Action someone takes: a content strategist gets a short list of pages flagged as "engagement risk" with the likely reason (e.g. thin content, mismatched search intent, poor formatting) instead of manually auditing hundreds of pages.

Cost of a wrong call: if we flag a healthy page as broken, the client wastes writer/editor hours on a rewrite that doesn't move the needle.
